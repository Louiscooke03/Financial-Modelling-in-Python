```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
from scipy.optimize import fsolve
from scipy.stats import mode
```

```python
def black_scholes_call(S, K, T, t, r, sigma, D):
    tau = T - t
    if tau <= 0:
        return max(S - K, 0)
    
    d1 = (np.log(S/K) + (r - D + 0.5*sigma**2)*tau) / (sigma*np.sqrt(tau))
    d2 = d1 - sigma*np.sqrt(tau)
    
    return S*np.exp(-D*tau)*norm.cdf(d1) - K*np.exp(-r*tau)*norm.cdf(d2)

K0 = 80
K1 = 100
K2 = 120
r = 0.05
sigma = 0.2
D = 0.02
T = 1.0
t = 0

S_values = np.linspace(50, 150, 100)
call_prices0 = [black_scholes_call(S, K0, T, t, r, sigma, D) for S in S_values]
call_prices1 = [black_scholes_call(S, K1, T, t, r, sigma, D) for S in S_values]
call_prices2 = [black_scholes_call(S, K2, T, t, r, sigma, D) for S in S_values]

plt.figure(figsize=(10, 6))
plt.plot(S_values, call_prices0, ls = '-', color = 'k', label = 'K = 80')
plt.plot(S_values, call_prices1, ls = '--', color = 'k', label = 'K = 100')
plt.plot(S_values, call_prices2, ls = '-.', color = 'k', label = 'K = 120')
plt.xlabel('S', fontsize=14)
plt.xticks([50, 60, 70, 80, 90, 100, 110, 120, 130, 140])
plt.ylabel('C(S, 0)', fontsize=14)
plt.grid(True)
plt.legend()
plt.show()
```

```python
def equations(vars, S, strike_price, maturity, init_t, C):
    r_init, sigma_init, D_init = vars
    eq1 = black_scholes_call(S[0], strike_price[0], maturity, init_t, r_init, sigma_init, D_init) - C[0]
    eq2 = black_scholes_call(S[1], strike_price[1], maturity, init_t, r_init, sigma_init, D_init) - C[1]
    eq3 = black_scholes_call(S[2], strike_price[2], maturity, init_t, r_init, sigma_init, D_init) - C[2]
    return [eq1, eq2, eq3]

S_varied = np.array([90, S_fixed, 130])
strike_price_fixed = np.array([K1, K1, K1])
maturity = 1
init_t = 0
C0 = black_scholes_call(S_varied[0], strike_price_fixed[0], maturity, init_t, 0.05, 0.2, 0.02)
C1 = black_scholes_call(S_varied[1], strike_price_fixed[1], maturity, init_t, 0.05, 0.2, 0.02)
C2 = black_scholes_call(S_varied[2], strike_price_fixed[2], maturity, init_t, 0.05, 0.2, 0.02)

initial_guess = [0.5, 0.5, 0.5]
solution0 = fsolve(equations, initial_guess, args=(S_varied, strike_price_fixed, maturity, init_t, [C0, C1, C2]))

print(f"The value of r that solves the system of equations is: {solution0[0]:.15f}")
print(f"The value of σ that solves the system of equations is: {solution0[1]:.15f}")
print(f"The value of D that solves the system of equations is: {solution0[2]:.15f}")
```

```python
def equations_solver(Stockprices, Strikeprices, Maturitytime, Initialtime, interestrate, volatility, dividend_yield):
    
    Calloptprem1 = black_scholes_call(Stockprices[0], Strikeprices[0], Maturitytime, Initialtime, interestrate, volatility, dividend_yield)
    Calloptprem2 = black_scholes_call(Stockprices[1], Strikeprices[1], Maturitytime, Initialtime, interestrate, volatility, dividend_yield)
    Calloptprem3 = black_scholes_call(Stockprices[2], Strikeprices[2], Maturitytime, Initialtime, interestrate, volatility, dividend_yield)

    solutions = []
    guesses = []
    errors = []
    for i in range(0, 800):
        r_star = np.random.lognormal(-1, 3)
        sigma_star = np.random.lognormal(-1, 3)
        D_star = np.random.lognormal(-1, 3)

        initialGuess = [r_star, sigma_star, D_star]
        solution = fsolve(equations, initialGuess, args=(Stockprices, Strikeprices, Maturitytime, Initialtime, [round(Calloptprem1, 1), round(Calloptprem2, 1), round(Calloptprem3, 1)]))
        
        error = np.array(solution) - np.array(initialGuess)
        if np.sum(error) != 0: # Case where fsolve() takes the guess as solution is invalid
            if np.abs(error[0]) < 0.05:
                if np.abs(error[1]) < 0.05:
                    if np.abs(error[2]) < 0.05:
                        solutions.append(solution)
                        guesses.append(initialGuess)
                        errors.append(error)
    return np.array(solutions), np.array(guesses), np.array(errors)

solver1, guess1, errors1 = equations_solver(S_varied, strike_price_fixed, 10, 1, 0.05, 0.2, 0.02)
```

```python
def optimal_solution(solver, guess, errors):

    rounded_arr = np.round(solver, 8)
    mode_row = mode(rounded_arr, axis=0, keepdims=False).mode

    matching_indices = np.where((np.round(solver, 8) == mode_row).all(axis=1))[0]
    abs_errors = np.sum(errors**2, axis=1)

    err_of_solns = []
    for i in matching_indices:
        err_of_soln = abs_errors[i]
        err_of_solns.append(err_of_soln)

    loc_of_min_err_soln = np.where(abs_errors == np.min(err_of_solns))[0][0]
    return mode_row, guess[loc_of_min_err_soln]

opt_soln1, opt_init_guess1 =  optimal_solution(solver1, guess1, errors1)
err_of_prediction1 = np.linalg.norm(np.array([0.05, 0.2, 0.02]) - opt_soln1)
print(f"The optimal solution is: {opt_soln1}")
print(f"The optimal initial guess is: {opt_init_guess1}")
print(f"The total error is: {err_of_prediction1}")
```

```python

from scipy.optimize import least_squares
def equations_tikh(vars, S, strike_price, maturity, init_t, market_prices, lambda_r, lambda_sigma, lambda_D, r_ref, sigma_ref, D_ref):
    r_init, sigma_init, D_init = vars
    eq1 = black_scholes_call(S[0], strike_price[0], maturity, init_t, r_init, sigma_init, D_init) - market_prices[0]
    eq2 = black_scholes_call(S[1], strike_price[1], maturity, init_t, r_init, sigma_init, D_init) - market_prices[1]
    eq3 = black_scholes_call(S[2], strike_price[2], maturity, init_t, r_init, sigma_init, D_init) - market_prices[2]
    residuals = np.array([eq1, eq2, eq3])
    
    reg_r = lambda_r * (r_init - r_ref)**2
    reg_sigma = lambda_sigma * (sigma_init - sigma_ref)**2
    reg_D = lambda_D * (D_init - D_ref)**2
    
    return np.concatenate([residuals, [np.sqrt(reg_r), np.sqrt(reg_sigma), np.sqrt(reg_D)]])
```

```python
result1 = least_squares(equations_tikh, x0=[0.1, 0.1, 0.1], bounds=([0, 0, 0], [1, 1, 1]), args=(np.array([637.1, 637.1, 637.1]), np.array([K200_2027, K205_2027, K210_2027]), T_2027, 0, np.array([C200_2027, C205_2027, C210_2027]), 100, 0.01, 0, 0.04, 0.25, 0.01))
lsq_err1 = np.sum(result1.fun[:3]**2)
rglr_err1 = np.sum(result1.fun[3:]**2)
print(f"The total least squares error is {lsq_err1}")
print(f"The total resularised error is {rglr_err1}")
print(f"The solution is {result1.x}")
```
